```python
# Cell 9 — Audit the current Proposed logs without making API calls
from collections import Counter
from pathlib import Path
import json

logs = sorted(Path('logs').glob('*__Variant_1_ErrorTrace__run*__groq.json'))
print('Completed Variant-1 Groq logs:', len(logs))

if logs:
    data = [json.loads(p.read_text(encoding='utf-8')) for p in logs]
    functions = sorted({d['function_id'] for d in data})
    by_function = Counter(d['function_id'] for d in data)
    print('Functions covered:', len(functions), functions)
    print('Runs per function:', dict(sorted(by_function.items())))
    print('Iteration counts:', Counter(d.get('iteration_count') for d in data))
    print('Mutation scores:', Counter(d.get('mutation_score_pct') for d in data))
    print('Logs with refinement detail:', sum(bool(d.get('iterations_detail')) for d in data))

    missing = []
    for p, d in zip(logs, data):
        for field in ('function_id','system_variant','iteration_count','total_tokens_used','mutation_score_pct'):
            if field not in d:
                missing.append((p.name, field))
    print('Required-field issues:', missing or 'none')
else:
    print('No Variant-1 Groq logs found yet.')
```
